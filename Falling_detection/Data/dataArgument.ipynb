```python
import pandas as pd
from sklearn.utils import shuffle
```

```python
df_fall = pd.read_csv("falling.csv", skiprows=1, names=["timestamp","accel_x","accel_y","accel_z","gyro_x","gyro_y","gyro_z","label"])
df_walk = pd.read_csv("walking.csv", skiprows=1, names=["timestamp","accel_x","accel_y","accel_z","gyro_x","gyro_y","gyro_z","label"])
df_sit = pd.read_csv("sitting.csv", skiprows=1, names=["timestamp","accel_x","accel_y","accel_z","gyro_x","gyro_y","gyro_z","label"])
df_stand = pd.read_csv("standing.csv", skiprows=1, names=["timestamp","accel_x","accel_y","accel_z","gyro_x","gyro_y","gyro_z","label"])
```

```python
df = pd.concat([df_fall, df_walk, df_sit, df_stand], ignore_index=True)
df = shuffle(df).reset_index(drop=True)
df.to_csv("all_data_clean.csv", index=False)
print("Combined and saved all_labeled_data.csv")
```
